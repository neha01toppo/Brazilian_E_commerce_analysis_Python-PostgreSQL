### customer_level table

```python
query='''with purchase_spent as(
    select order_id,count(order_id)as quantity,sum(price)as spent_price
	from order_item
	group by 1)

select c.customer_id,c.customer_unique_id,c.customer_city,c.customer_state,
       o.order_status,
       p.quantity,p.spent_price
from customers c
left join orders o on c.customer_id=o.customer_id
left join purchase_spent p on o.order_id=p.order_id'''
```

```python
df=pd.read_sql_query('''with purchase_spent as(
    select order_id,count(order_id)as quantity,sum(price)as spent_price
	from order_item
	group by 1)

select c.customer_id,c.customer_unique_id,c.customer_city,c.customer_state,
       o.order_status,
       p.quantity,p.spent_price
from customers c
left join orders o on c.customer_id=o.customer_id
left join purchase_spent p on o.order_id=p.order_id''',engine)
```

```python
df['spent_price']=df['spent_price'].round(2)
df['quantity']=df['quantity'].round(0)
display(df)
```

## Product_level data load from pgadmin

```python
query="""with order_product as
  (select product_id,order_id,
        round(sum(price)::numeric,2)as product_price
  from order_item
  group by 1,2),

review_product as(
     select o.order_id,o.product_id,r.review_score
	 from order_item o
	  join order_review r on r.order_id=o.order_id)

select o.product_id,o.order_id,o.product_price,
       r.review_score,
	   p.product_category
from order_product o
left join review_product r on o.order_id=r.order_id
 join products p on o.product_id=p.product_id;
       """
```

```python
df_p=pd.read_sql_query('''with order_product as
  (select product_id,order_id,
        round(sum(price)::numeric,2)as product_price
  from order_item
  group by 1,2),

review_product as(
     select o.order_id,o.product_id,r.review_score
	 from order_item o
	  join order_review r on r.order_id=o.order_id)

select o.product_id,o.order_id,o.product_price,
       r.review_score,
	   p.product_category
from order_product o
left join review_product r on o.order_id=r.order_id
 join products p on o.product_id=p.product_id;
       ''',engine)
```

```python
df_p
```

```python
df_p.info()
```

```python
df_p.isnull().sum()
```

```python
null_r=df_p[df_p.isnull().any(axis=1)]
null_r
```

```python
df_p=df_p.dropna()
```

```python
df_p['review_score'].unique()
```

```python
map_review={
    1:'very bad',
    2:'bad',
    3:'average',
    4:'good',
    5:'very good'
}

df_p['review_label']=df_p['review_score'].map(map_review)
df_p
```

```python
total_rev= df_p.groupby(['product_id','product_category']).agg({
    'product_price':'sum',
    'review_score':'count'
})

total_rev=total_rev.sort_values(['review_score'],ascending=False)
total_rev
```

## Load in postgre.

```python
df_p.to_sql(
    name='product_level',
        con=engine,
    if_exists='replace',
    index= False
)
```
